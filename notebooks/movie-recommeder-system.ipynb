Preprocessing Genres 

```python
movies.iloc[0].genres
```

```python
# this is string of list so we need to convert it to list so we need the ast module
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
# removing ids and making list of the genres 

def convert(obj):
    genreList=[]
    for i in ast.literal_eval(obj):
        genreList.append(i['name'])
    return genreList
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python

movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head(3)
```

Getting the top 4 cast 

```python
# movies['cast'][0]
def getCast(obj):
    castList = []
    counter=0
    for i in ast.literal_eval(obj):
        if counter!= 4:
            castList.append(i['name'])
            counter+=1
        else:
            break
    return castList
```

```python
movies['cast'] = movies['cast'].apply(getCast)
```

```python
movies.head(3)
```

```python
# movies['crew'][0]
```

```python
# movies['cast'][0]
def getDirector(obj):
    director = []
    counter=0
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            director.append(i['name'])
            break
    return director
```

```python
movies['crew'] =movies['crew'].apply(getDirector)
```

```python
movies.head()
```

```python
movies['overview'][0] # this is string so converting to list so that it can be easily concatenated to other tags 

movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

Removing the spaces 

```python
# 'Sam Worthington' -> 'SamWorthington' this is neccesssary to remove the space as the model might confuse one sam with other 
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['tags']= movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.head(1)
```

```python
final_movies = movies[['movie_id','title','tags']].copy()
```

```python
final_movies.head(1)
```

```python
# converting the tag to strings

final_movies['tags'] = final_movies['tags'].apply(lambda x: " ".join(x))
```

```python
final_movies.head(1)
```

```python
final_movies['tags'][0]
```

```python
final_movies['tags'] = final_movies['tags'].apply(lambda x:x.lower())
```

Vectorization 

```python
from sklearn.feature_extraction.text import CountVectorizer 
cv= CountVectorizer(max_features=5000 , stop_words='english')  # Vectorize based on the frequencies/ count of the 5000 most common word s 
#  Stop words like is are etc
```

```python
vectors = cv.fit_transform(final_movies['tags']).toarray()
```

Stemming : 
[ loving, loved, love] to [love , love, love ]

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
ps.stem("Having")
ps.stem("Loving")
```

```python
# helper function 
def stem(text):
    y=[]

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
stem('in the 22nd century, a paraplegic marine is dispatched to the moon pandora on a unique mission, but becomes torn between following orders and protecting an alien civilization. action adventure fantasy sciencefiction cultureclash future spacewar spacecolony society spacetravel futuristic romance space alien tribe alienplanet cgi marine soldier battle loveaffair antiwar powerrelations mindandsoul 3d samworthington zoesaldana sigourneyweaver stephenlang jamescameron')
```

```python
final_movies['tags'] =final_movies['tags'].apply(stem)
```

```python
vectors = cv.fit_transform(final_movies['tags']).toarray()
vectors.shape
```

```python
# print(cv.get_feature_names_out()) 


# print(len(cv.get_feature_names_out()))

# print(list(cv.get_feature_names_out())) # to avoid truncation 
```

Now the word with same meaning is not duplicated in vectors 

Using the cosine distance . Not Euclidian distance as Euclidean distance is not good for the high dimenstional data As we have 5000 D data we can use the cosine distance 

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors) # this calculate the distance between the each movies with each other and give similarity
```

```python
similarity[0] # similarity of first movie Avatar with all movies 
```

We can see 1 which is similarity of the first movies with itself which will be 1 and 0.08 is similarity between first and second movies 

```python
sorted(list(enumerate(similarity[0])), reverse=True , key=lambda x:x[1])[1:6] # getting the top 5 similar movies with its index 
# sorted(similarity[0])[-10:-1]
```

```python
def recommend(movie):
    movie_index = final_movies[final_movies['title'] == movie].index[0] # get index based on the title 
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(similarity[0])), reverse=True , key=lambda x:x[1])[1:6]

    for i in movies_list:
        # print(i[0])
        print(final_movies.iloc[i[0]].title)


```

```python
recommend('Avatar')
```
